```python
import json
from pathlib import Path

results_path = Path("results/sample_run.json")

with results_path.open("r", encoding="utf-8") as f:
    results = json.load(f)

print("Total episodes:", results["n_episodes"])
print("Calibration safe episodes:", results["n_calibration_safe"])
print("Test episodes:", results["n_test_episodes"])
```

```python
conditions = results["conditions"]

comparison = []

for name, data in conditions.items():
    metrics = data["overall"]
    comparison.append({
        "Monitor": name,
        "Threshold": data["threshold"],
        "Recall": metrics["recall_at_fpr"],
        "Precision": metrics["precision"],
        "F1": metrics["f1"],
        "AUROC": metrics["auroc"],
        "FPR": metrics["fpr"],
    })

comparison
```

```python
tests = results["tests"]

for test_name, test_result in tests.items():
    print(f"\n{test_name}")
    for metric, value in test_result.items():
        print(f"  {metric}: {value}")
```

## Preliminary Results and Limitations

The sample benchmark contains 10 demonstration episodes, including 6 unsafe episodes. The output-only monitor achieved a recall of 0.0, while the reasoning-trace and output monitor achieved a recall of 1.0.

For RQ1, the category homogeneity test was not testable because there were no missed unsafe cases across categories. For RQ2, the one-sided exact binomial test did not provide sufficient evidence that recall exceeded 0.70 (p = 0.1176). For RQ3, the one-sided exact McNemar test returned p = 0.015625, and the paired bootstrap estimated a recall difference of 1.0.

These results are preliminary smoke-test results from a very small demonstration dataset. They do not establish that reasoning traces improve unsafe-behavior detection in general. A larger evaluation dataset and the intended monitor configuration are required before drawing research conclusions.
